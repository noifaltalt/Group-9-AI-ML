```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
FILE_PATH = "../data/15-data.csv"
df = pd.read_csv(FILE_PATH)
```

```python
# Select hyperparameter features used as model inputs
features = ["Step", "learning_rate", "num_epoch", "batch_size", "buffer_size", "beta", "epsilon", "lambd"]
X = df[features]

# Identify numerical features (all selected features are expected to be numeric)
numeric_features = X.select_dtypes(include=[np.number]).columns.tolist()

# Define preprocessing pipeline:
# - Impute missing values using the median
# - Standardize features to zero mean and unit variance
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Apply preprocessing features using the defined pipeline
# All non-specified columns are dropped to ensure the model receives
# only cleaned and standardized numerical inputs
preprocess = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features)
    ],
    remainder="drop"
)
```

```python
# Initialize a Random Forest Regressor with 500 trees
# - n_jobs=-1 enables parallel training using all available CPU cores
model = RandomForestRegressor(
    n_estimators=500,
    random_state=42,
    n_jobs=-1
)

# Build a pipeline that combines preprocessing and model training
pipe = Pipeline(steps=[
    ("preprocess", preprocess),
    ("model", model)
])
```

```python
# Predicting the Entropy of Last Step
entropy_y = df["entropy"]
X_train, X_test, y_train, y_test = train_test_split(
    X,entropy_y,test_size=0.2,random_state=42,stratify=None
)
pipe.fit(X_train, y_train)

pred = pipe.predict(X_test)
rmse = mean_squared_error(y_test, pred)
mae = mean_absolute_error(y_test, pred)
r2 = r2_score(y_test, pred)

print("----- Entropy -----")
print(f"RMSE: {rmse:.4f}")
print(f"MAE : {mae:.4f}")
print(f"R^2 : {r2:.4f}")
```

```python
# Predicting the Episode Length of Last Step
episodeLen_y = df["episode_length"]
X_train, X_test, y_train, y_test = train_test_split(
    X,episodeLen_y,test_size=0.2,random_state=42,stratify=None
)
pipe.fit(X_train, y_train)

pred = pipe.predict(X_test)
rmse = mean_squared_error(y_test, pred)
mae = mean_absolute_error(y_test, pred)
r2 = r2_score(y_test, pred)

print("----- Episode Length -----")
print(f"RMSE: {rmse:.4f}")
print(f"MAE : {mae:.4f}")
print(f"R^2 : {r2:.4f}")
```

```python
# Predicting the Cumulative Reward of Last Step
episodeLen_y = df["cumulative_reward"]
X_train, X_test, y_train, y_test = train_test_split(
    X,episodeLen_y,test_size=0.2,random_state=42,stratify=None
)
pipe.fit(X_train, y_train)

pred = pipe.predict(X_test)
rmse = mean_squared_error(y_test, pred)
mae = mean_absolute_error(y_test, pred)
r2 = r2_score(y_test, pred)

print("----- Cumulative Reward -----")
print(f"RMSE: {rmse:.4f}")
print(f"MAE : {mae:.4f}")
print(f"R^2 : {r2:.4f}")
```
